# Phase 6c: ablation controls (Qwen2.5-7B-Instruct, bf16)

**A100**, then `Runtime → Run all`. Needs Phase 5/6/6b outputs on Drive.

Phase 6b: removing v at every layer from the pushback turn on erased the senior − junior gap (+4.3 → −1.6 points),
mostly by lowering senior flips, while 2 random directions did ~nothing. But random directions are a weak control for
ablation (a random direction barely overlaps the activations, so removing it removes little), and the shuffled
vector, which *is* data-shaped, moved gaps a lot in other settings. This run, same setting (all layers, from pushback):

1. **Fairer controls:** the shuffled vector + the 6 per-pair shuffled vectors (data-shaped, no consistent senior/junior sign),
   and 8 more random directions (10 total).
2. **Where:** ablate v (and the shuffled vector) in layer bands 0–9, 10–19, 20–27 only.
